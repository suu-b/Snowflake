# EDA of the generated Synthetic Data

```python
import pandas as pd
import numpy as np 
import matplotlib.pyplot as plt 
import seaborn as sb
```

```python
df = pd.read_csv('./data/data.csv')
pd.to_datetime(df['as_of_date'])
```

```python
date_counts = df['as_of_date'].value_counts().sort_index()
if date_counts.nunique() == 1:
    print(f"Every date has exactly {date_counts.iloc[0]} rows.")
    print(f"Total dates: {len(date_counts)}")
    print(f"Translates to roughly {len(date_counts) // 365} years.")
    
else:
    print("Inconsistent row counts.")
    
plt.figure(figsize=(12, 5))
date_counts.plot(kind='line', color='teal', linewidth=2)
```

```python
df.plot(x='as_of_date', y='value', figsize=(12, 5))
```

```python
# This is an example of multiple time series
# The graph shows 5 time series (5 specific combinations of (key1, key2, key3)) 

df_wide = df.pivot(index='as_of_date', 
                   columns=['key_1', 'key_2', 'key_3'], 
                   values='value')
df_wide.iloc[:, :5].plot(figsize=(15, 7), linewidth=1)

plt.title('5 Individual Time Series (Wide Format View)', fontsize=14)
plt.ylabel('Value')
plt.legend(title='Key Combinations', bbox_to_anchor=(1.05, 1), loc='upper left')
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```
