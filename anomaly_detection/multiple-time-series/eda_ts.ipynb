# EDA of TimeSynth Data

```python
import pandas as pd 
import matplotlib.pyplot as plt 
import seaborn as sb
import numpy as np
```

```python
df = pd.read_csv("./data/ts_data.csv")
```

```python
import plotly.express as px

target_series = df[(df['key_1'] == 'key_1_a') & 
                   (df['key_2'] == 'key_2_a') & 
                   (df['key_3'] == 'key_3_a')]

fig = px.line(target_series, x='as_of_date', y='value', 
              title='Synthetic Time Series (Trend + Weekly Seasonality)')
fig.show()
```

```python
jan_data = target_series[
    (target_series['as_of_date'] >= '2023-01-01') & 
    (target_series['as_of_date'] <= '2023-01-31')
]
jan_data['as_of_date'] = pd.to_datetime(jan_data['as_of_date'])
fig = px.line(
    jan_data, 
    x='as_of_date', 
    y='value', 
    title='Seasonality Check: January 2023 (Series: key_1_a | key_2_a | key_3_a)',
    markers=True,        
    template='plotly_white'
)

for date in jan_data[jan_data['as_of_date'].dt.dayofweek == 0]['as_of_date']:
    fig.add_vline(x=date.timestamp() * 1000, line_dash="dash", line_color="red", opacity=0.3)
fig.show()
```

```python
jan_data = target_series[
    (target_series['as_of_date'] >= '2025-12-01') & 
    (target_series['as_of_date'] <= '2025-12-31')
]
jan_data['as_of_date'] = pd.to_datetime(jan_data['as_of_date'])
fig = px.line(
    jan_data, 
    x='as_of_date', 
    y='value', 
    title='Seasonality Check: January 2025 (Series: key_1_a | key_2_a | key_3_a)',
    markers=True,        
    template='plotly_white'
)

for date in jan_data[jan_data['as_of_date'].dt.dayofweek == 0]['as_of_date']:
    fig.add_vline(x=date.timestamp() * 1000, line_dash="dash", line_color="red", opacity=0.3)
fig.show()
```
